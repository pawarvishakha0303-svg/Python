```python
#  Find sum of 1 to n numbers

n = int(input("Enter a number: "))
sum = 0

for i in range(1, n + 1):
    sum = sum + i

print("Sum =", sum)
```

```python
# Factorial of Number

n = int(input("Enter a number: "))
fact = 1

for i in range(1, n + 1):
    fact = fact * i

print("Factorial =", fact)
```

```python
# Digit count in number

n = input("Enter a number: ")
count = 0

for i in n:
    count += 1

print("Digits =", count)
```

```python
# Reverse String

text = input("Enter a string: ")

for i in range(len(text) - 1, -1, -1):
    print(text[i], end="")
```

```python
# Sum of Even numbers

n = int(input("Enter a number: "))
sum = 0

for i in range(2, n + 1, 2):
    sum += i

print("Sum =", sum)
```

```python
# Sum of Odd numbers

n = int(input("Enter a number: "))
sum = 0

for i in range(1, n + 1, 2):
    sum += i

print("Sum =", sum)
```

```python
# Number Pattern Printing

n = int(input("Enter rows: "))

for i in range(1, n + 1):
    for j in range(1, i + 1):
        print(j, end="")
    print()
```

```python
# Alphabet pattern 

n = int(input("Enter rows: "))

for i in range(1, n + 1):
    for j in range(i):
        print(chr(64 + i), end=" ")
    print()
```

```python
# Pyramid

n = int(input("Enter rows: "))

for i in range(1, n + 1):
    print(" " * (n - i) + "*" * (2 * i - 1))
```

```python
# Floyds Triangle

n = int(input("Enter rows: "))
num = 1

for i in range(1, n + 1):
    for j in range(i):
        print(num, end=" ")
        num += 1
    print()
```
