#  With Arguments, No Return Value 

```python
# Program 1: Addition

def add(a, b): 
    print("Sum =", a + b) 
add(10, 20)
```

```python
# Program 2: Subtraction

def subtract(a, b): 
    print("Difference =", a - b) 
subtract(20, 10)
```

```python
# Program 3: Multiplication 

def multiply(a, b):
    print("Product =", a * b) 
multiply(5, 6)
```

```python
# Program 4: Division 

def divide(a, b):
    if b != 0:
        print("Quotient =", a / b) 
    else: print("Cannot divide by zero") 
divide(20, 4) 
```

```python
# Program 5: Square

def square(n): 
    print("Square =", n*n) 
square(7)
```

```python
#  Program 6: Cube 

def cube(n): 
    print("Cube =", n**3) 
cube(4) 
```

```python
# Program 7: Even or Odd 

def even_odd(n): 
    if n%2==0: 
        print("Even") 
    else: 
        print("Odd") 
even_odd(15)
```

```python
# Program 8: Maximum

def maximum(a,b): 
    print("Maximum =", a if a>b else b) 
maximum(12,8) 
```

```python
#Program 9: Minimum 

def minimum(a,b):
    print("Minimum =", a if a<b else b)
minimum(12,8)
```

```python
#Program 10: Swap Numbers

def swap(a,b): 
    a,b=b,a 
    print(a,b) 
swap(10,20)
```
