```python
#Multiplication Table

num=int(input("Enter a number:"))
i=1
while(i<=10):
    print(i*num)
    i=i+1
```

```python
# Factorial of number

n1=int(input("Enter a number:"))
fact=1
i=1
while(i<=n1):
    fact=fact*i
    i=i+1
print(fact)
```

```python
# Sum of first n numbers

n=int(input("Enter a number:"))
i=1
total=0

while i<=n:
    total=total+i
    i=i+1
print("Sum",total)
```

```python
# Product of first n numbers

n=int(input("Enter a number:"))
i=1
product=1

while i<=n:
    product=product*i
    i=i+1
print("Product",product)
```

```python
# Count of digits

num=int(input("Enter a number:"))
count=0

while num>0:
    count=count+1
    num=num//10
print("Number of digits=",count)
```

```python
# Fibonacci series

n = int(input("Enter number:"))
a=0
b=1
count=1

while count<=n:
    print(a)
    c=a+b
    a=b
    b=c
    count=count+1

```

```python
# Perfect number 

num=int(input("Enter a number:"))
i=1
sum=0

while i<num:
    if num%i==0:
        sum=sum+i
    i=i+1
if sum==num:
    print("Perfect number")
else:
    print("Not Perfect number")

```

```python
# Reverse number

num=int(input("Enter a number:"))
rev=0

while num>0:
    digit=num%10
    rev=rev*10+digit
    num=num//10
print("Reverse No=",rev)
```

```python
# Prime number

num=int(input("Enter a number:"))
i=2
while i<num:
    if num%i==0:
        print("Not Prime")
        break
    i=i+1
else:
    print("Prime")
```

```python
# Palindrome number

num=int(input("Enter a number:"))
temp=num
reverse=0

while num>0:
    digit=num%10
    reverse=reverse*10+digit
    num=num//10

if temp==reverse:
    print("Palindrome number")
else:
    print("Not Palindrome number")
```
