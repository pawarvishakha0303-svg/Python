```python
# Union of two set

a={1,2,3,4,5}
b={6,7,8,9,4}

print(a.union(b))
```

```python
# Intersection of two set

a={1,2,3,4,5}
b={6,7,8,9,4}

print(a.symmetric_difference(b))
```

```python
# Symmetric difference of two set

a={1,2,3,4,5}
b={6,7,8,9,4}

print(a.intersection(b))
```

```python
# Remove duplicate elements

l={1,2,2,3,4,5,4,6,3,7,2,1}
s=set(l)
print(s)
```

```python
# Find difference between multiple sets

a={1,2,3,4,5}
b={2,3,6}
c={5}
print(a-b-c)
```

```python
# Find unique elements

a={51,56,89,34,29,66}
b={43,73,99,66,51,38}

print("Common:",a & b)
```

```python
# Find maximum Common element

a={10,20,30,40,50}
b={10,30,50,70,90}
common=a & b

print(max(common))
```

```python
# Remove common element fromtwo sets

a={2,4,6,8,10}
b={12,14,2,8,16}

common= a & b
a=a-common
b=b-common

print("Set A:",a)
print("B:",b)
```

```python
# Merge multiple sets

a={1,2,3}
b={4,5,6}
c={7,8,9}
d={10}

result= a | b | c 
print(result)
```
