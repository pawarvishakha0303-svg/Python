```python
# Store Student Data

name = ["Ram", "Siya", "Neha"]
marks = (85, 90, 88)
subjects = {"Python", "Java", "C"}
student = {"Name": "Ram", "Marks": 85}

print(name)
print(marks)
print(subjects)
print(student)
```

```python
# Find Length

l = [10, 20, 30]
t = (1, 2, 3, 4)
s = {5, 6, 7}
d = {"a": 1, "b": 2}

print(len(l))
print(len(t))
print(len(s))
print(len(d))
```

```python
# Add Elements

l = [10, 20]
l.append(30)

s = {1, 2}
s.add(3)

d = {"Name": "Ram"}
d["Age"] = 20

print(l)
print(s)
print(d)
```

```python
# Access Elements

l = [10, 20, 30]
t = (40, 50, 60)
d = {"Name": "Ram"}

print(l[0])
print(t[1])
print(d["Name"])
```

```python
# Remove Elements

l = [10, 20, 30]
l.remove(20)

s = {1, 2, 3}
s.remove(2)

d = {"a": 1, "b": 2}
d.pop("a")

print(l)
print(s)
print(d)
```

```python
# Convert List to Tuple, Set, and Dictionary

l = [10, 20, 30]

t = tuple(l)
s = set(l)
d = dict(enumerate(l))

print(t)
print(s)
print(d)
```

```python
# Loop Through Collections

l = [10, 20]
t = (30, 40)
s = {50, 60}
d = {"a": 1, "b": 2}

for i in l:
    print(i)

for i in t:
    print(i)

for i in s:
    print(i)

for k, v in d.items():
    print(k, v)
```

```python
# Find Maximum

l = [10, 50, 30]
t = (40, 70, 20)
s = {5, 15, 25}
d = {"a": 10, "b": 30}

print(max(l))
print(max(t))
print(max(s))
print(max(d.values()))
```

```python
# Find Sum

l = [10, 20, 30]
t = (40, 50)
s = {60, 70}
d = {"a": 10, "b": 20}

print(sum(l))
print(sum(t))
print(sum(s))
print(sum(d.values()))
```
