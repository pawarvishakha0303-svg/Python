# Part 1: Basic String Programs (Programs 1–10)





```python
# Program 1: Print a String

s = "Hello, World!"
print(s)

```

```python
# Program 2: Input and Display a String

s = input("Enter a string: ")
print("You entered:", s)
```

```python
# Program 3: Find Length of a String

s = input("Enter a string: ")
print("Length:", len(s))
```

```python
# Program 4: Convert to Uppercase

s = input("Enter a string: ")
print(s.upper())
```

```python
# Program 5: Convert to Lowercase

s = input("Enter a string: ")
print(s.lower())
```

```python
# Program 6: Reverse a String

s = input("Enter a string: ")
print("Reverse:", s[::-1])
```

```python
# Program 7: Check Palindrome

s = input("Enter a string: ")
if s == s[::-1]:
    print("Palindrome")
else:
    print("Not Palindrome")
    
```

```python
    
# Program 8: Count Vowels

s = input("Enter a string: ")
count = 0

for ch in s.lower():
    if ch in "aeiou":
        count += 1

print("Vowels:", count)
```

```python
# Program 9: Count Consonants

s = input("Enter a string: ")
count = 0

for ch in s.lower():
    if ch.isalpha() and ch not in "aeiou":
        count += 1

print("Consonants:", count)
```

```python
# Program 10: Count Digits, Alphabets and Special Characters

s = input("Enter a string: ")

letters = digits = special = 0

for ch in s:
    if ch.isalpha():
        letters += 1
    elif ch.isdigit():
        digits += 1
    else:
        special += 1

print("Letters:", letters)
print("Digits:", digits)
print("Special Characters:", special)
```
