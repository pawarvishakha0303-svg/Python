Nested_If Statement:
A nested if means an if statement inside another if statement. It is used when you want to check a second condition only if the first condition is true.

```python
# Check Voting Eligibility

age= int(input("Enter your age"))
if age>=18:
    if age>=21:
        print("Eligible for vote and drive vehicles")
    else:
        print("Eligible for vote")
else:
    print("Not eligiblr for vote")
```

```python
# Check Divisibiity 
num = int(input("Enter a number: "))

if num % 2 == 0:
    if num % 4 == 0:
        print("Divisible by 2 and 4")
    else:
        print("Divisible by 2 only")
else:
    print("Not divisible by 2")
```

```python
# Check positive odd or Even number

num = int(input("Enter a number: "))

if num > 0:
    if num % 2 == 0:
        print("Positive Even Number")
    else:
        print("Positive Odd Number")
else:
    print("Negative Number")
```

```python
# Check Employee bonus

salary = int(input("Enter salary: "))
experience = int(input("Enter years of experience: "))

if experience >= 3:
    if salary >30000:
        print("Bonus: 20%")
    else:
        print("Bonus: 10%")
else:
    print("No Bonus")
```

```python
# Username and Password

username = input("Enter username: ")
password = input("Enter password: ")

if username == "admin":
    if password == "1234":
        print("Login Successful")
    else:
        print("Wrong Password")
else:
    print("Invalid Username")
```

```python
# Scholarship

marks = int(input("Enter marks: "))
income = int(input("Enter family income: "))

if marks >= 75:
    if income < 200000:
        print("Scholarship Approved")
    else:
        print("Income Too High")
else:
    print("Marks Not Enough")
```

```python
# Driving License

age = int(input("Enter age: "))
test = input("Pass driving test? (yes/no): ")

if age >= 18:
    if test == "yes":
        print("License Approved")
    else:
        print("Pass the Driving Test")
else:
    print("Not Eligible")
```

```python
# Admission

per = float(input("Enter percentage: "))
doc = input("Documents submitted? (yes/no): ")

if per >= 60:
    if doc == "yes":
        print("Admission Confirmed")
    else:
        print("Submit Documents")
else:
    print("Admission Rejected")
```

```python
# Online Shopping

amount = int(input("Enter purchase amount: "))
member = input("Premium member? (yes/no): ")

if amount >= 1000:
    if member == "yes":
        print("20% Discount")
    else:
        print("10% Discount")
else:
    print("No Discount")
```

```python
# Bank Loan

age = int(input("Enter age: "))
salary = int(input("Enter salary: "))

if age >= 21:
    if salary >= 25000:
        print("Loan Approved")
    else:
        print("Salary Too Low")
else:
    print("Not Eligible")
```
