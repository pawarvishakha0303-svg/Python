```python
# Program 1: Return Greeting 

def greeting(): 
    return 'Hello, World!' 
print(greeting())
```

```python
# Program 2: Return Number

def get_number():
    return 100 
print(get_number())
```

```python
# Program 3: Return Current Year

from datetime import datetime
def current_year():
    return datetime.now().year
print(current_year())
```

```python
# Program 4: Return Square def square(): n=7 return n*n print(square())
```

```python
# Program 5: Return Cube

def cube(): 
    n=4 
    return n**3
print(cube())
```

```python
# Program 6: Return Even Numbers

def even_numbers():
    return [i for i in range(2,21,2)] 
print(even_numbers())
```

```python
# Program 7: Return Odd Numbers

def odd_numbers(): 
    return [i for i in range(1,20,2)]
print(odd_numbers())
```

```python
# Program 8: Return Alphabet

def alphabet(): 
    return ' '.join(chr(i) for i in range(65,91)) 
print(alphabet())
```

```python
# Program 9: Return List

def get_list():
    return [10,20,30]
print(get_list())
```

```python
# Program 10: Return Tuple 

def get_tuple():
    return (1,2,3) 
print(get_tuple())
```
