# filter() :
The filter() function is used to select elements from a list or other iterable based on a condition

```python
# Program 1 : Filter even numbers

numbers = [1, 2, 3, 4, 5, 6]

result = filter(lambda x: x % 2 == 0, numbers)

print(list(result))
```

```python
# Program 2 : Filter odd numbers

numbers = [1, 2, 3, 4, 5, 6]

result = filter(lambda x: x % 2 != 0, numbers)

print(list(result))
```

```python
# Program 3 : Filter numbers greater than 10

numbers = [5, 12, 8, 20, 15]

result = filter(lambda x: x > 10, numbers)

print(list(result))
```

```python
# Program 4 : Filter positive numbers

numbers = [-5, 10, -2, 8, 0]

result = filter(lambda x: x > 0, numbers)

print(list(result))
```

```python
# Program 5 : Filter negative numbers

numbers = [-5, 10, -2, 8, 0]

result = filter(lambda x: x < 0, numbers)

print(list(result))
```

```python
# Program 6 : Filter names starting with A

names = ["Amit", "Riya", "Anita", "Rahul"]

result = filter(lambda x: x.startswith("A"), names)

print(list(result))
```

```python
# Program 7 : Filter words with more than 5 characters

words = ["Python", "Java", "Programming", "Code"]

result = filter(lambda x: len(x) > 5, words)

print(list(result))
```

```python
# Program 8 : Filter numbers between 10 and 30

numbers = [5, 10, 15, 25, 35, 40]

result = filter(lambda x: 10 <= x <= 30, numbers)

print(list(result))
```

```python
# Program 9 :Filter numbers divisible by 5

numbers = [10, 12, 15, 22, 25, 30]

result = filter(lambda x: x % 5 == 0, numbers)

print(list(result))
```

```python
# Program 10 : Filter names with length greater than 4

names = ["Raj", "Vishakha", "Amit", "Priya"]

result = filter(lambda x: len(x) > 4, names)

print(list(result))
```
