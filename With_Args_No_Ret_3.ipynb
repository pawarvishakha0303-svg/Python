# With Arguments, No Return Value 

```python
# Program 1: Simple Interest

def simple_interest(p,r,t): 
    si=(p*r*t)/100 
    print('Simple Interest =',si) 
simple_interest(1000,5,2) 
```

```python
# Program 2: Area of Circle

def area_circle(r): 
    print('Area =',3.14*r*r)
area_circle(7)
```

```python
# Program 3: Area of Rectangle 

def area_rectangle(l,b): 
    print('Area =',l*b)
area_rectangle(10,5) 
```

```python
#Program 4: Temperature Converter

def c_to_f(c):
    print('F =',(c*9/5)+32) 
c_to_f(37) 
```

```python
# Program 5: Voting Eligibility 

def vote(age): 
    print('Eligible' if age>=18 else 'Not Eligible') 
vote(20)
```

```python
# Program 6: Student Details 

def student(name,age):
    print(name,age) 
student('Vishakha',20) 
```

```python
# Program 7: Employee Details

def employee(name,salary):
    print(name,salary) 
employee('Amit',50000)
```

```python
# Program 8: Bank Account

def account(name,balance): 
    print(name,balance) 
account('Riya',10000)
```

```python
# Program 9: Calculator

def add(a,b):
    print(a+b) 
add(10,20)
```

```python
# Program 10: Shopping Bill

def bill(price,qty): 
    print('Total =',price*qty) 
bill(250,4)
```
