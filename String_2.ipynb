# Part 2: Intermediate String Programs (Programs 11–20)


```python
# Program 11: Count Words in a String

s = input("Enter a sentence: ")

words = s.split()

print("Number of words:", len(words))
```

```python
# Program 12: Count Frequency of Each Character

s = input("Enter a string: ")

for ch in set(s):
    
    print(ch, ":", s.count(ch))
```

```python
 # Program 13: Remove All Spaces from a String

s = input("Enter a string: ")

result = s.replace(" ", "")

print("String without spaces:", result)
```

```python
# Program 14: Replace a Word in a String

s = input("Enter a sentence: ")

old = input("Enter word to replace: ")
new = input("Enter new word: ")

print("Updated string:", s.replace(old, new))
```

```python
# Program 15: Check Whether Two Strings are Anagrams

s1 = input("Enter first string: ")
s2 = input("Enter second string: ")

if sorted(s1.lower()) == sorted(s2.lower()):
    print("Anagram")
else:
    print("Not Anagram")
```

```python
# Program 16: Find the First Non-Repeated Character

s = input("Enter a string: ")

for ch in s:
    if s.count(ch) == 1:
        print("First non-repeated character:", ch)
        break
else:
    print("No unique character found")
    
```

```python
# Program 17: Count Uppercase and Lowercase Letters

s = input("Enter a string: ")

upper = 0
lower = 0

for ch in s:
    if ch.isupper():
        upper += 1
    elif ch.islower():
        lower += 1

print("Uppercase letters:", upper)
print("Lowercase letters:", lower)
```

```python
# Program 18: Find Duplicate Characters

s = input("Enter a string: ")

printed = ""

for ch in s:
    if s.count(ch) > 1 and ch not in printed:
        print(ch)
        printed += ch
        
```

```python
# Program 19: Find the Longest Word in a Sentence

s = input("Enter a sentence: ")

words = s.split()

longest = max(words, key=len)

print("Longest word:", longest)
```

```python
# Program 20: Sort Characters in Alphabetical Order

s = input("Enter a string: ")

result = "".join(sorted(s))

print("Sorted string:", result)
```
