```python

# Program 1: Print numbers 1 to 100

def program():
    for i in range(1, 25):
        print(i)

program()


 
```

```python
# Program 2: Print numbers 100 to 1

def num():
    for i in range(100, 0, -1):
        print(i)

num()
```

```python
# Program 3: Print even numbers 1 to 100

def even():
    for i in range(2, 101, 2):
        print(i)

even()
```

```python

# Program 4: Print odd numbers 1 to 100

def odd():
    for i in range(1, 101, 2):
        print(i)

odd()
```

```python
# Program 5: Print multiples of 10

def program():
    for i in range(10, 101, 10):
        print(i)

program()
```

```python

# Program 6: Print squares 1-20

def square():
    for i in range(1, 21):
        print(i, "=", i ** 2)

square()
```

```python
# Program 7: Print cubes 1-20

def cube():
    for i in range(1, 21):
        print(i, "=", i ** 3)

cube()
```

```python

# Program 8: Print tables 1-10

def table():
    for i in range(1, 11):
        print("Table of", i)
        for j in range(1, 11):
            print(i, "x", j, "=", i * j)

table()
```

```python
# Program 9: Sum 1-100

def sum():
    total = 0

    for i in range(1, 101):
        total += i

    print("Sum =", total)
sum()
```

```python
# Program 10: Powers of 2

def pow():
    for i in range(0, 11):
        print("2 ^", i, "=", 2 ** i)

pow()
```
