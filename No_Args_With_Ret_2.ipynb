```python

# Program 11: Return Dictionary 

def get_dict(): 
    return {'name':'Vishakha','age':20} 
print(get_dict())
```

```python
#Program 12: Return Set 

def get_set():
    return {10,20,30}
print(get_set())
```

```python
# Program 13: Return String 

def get_string(): 
    return 'Python' 
print(get_string()) 
```

```python
# Program 14: Reverse String 
def reverse():
    s='Python' 
    return s[::-1] 
print(reverse())
```

```python
# Program 15: Palindrome

def palindrome():
    s='madam' 
    return s==s[::-1]
print(palindrome())
```

```python
# Program 16: Fibonacci List 
def fibonacci():
    a,b=0,1;res=[] 
    for _ in range(10):
        res.append(a);a,b=b,a+b 
    return res
print(fibonacci())
```

```python
# Program 17: Prime Numbers

def primes(): 
    r=[] 
    for n in range(2,21):
        if all(n%i for i in range(2,n)):
            r.append(n)
            return r
print(primes())
```

```python
# Program 18: Factorial

def factorial():
    f=1 
    for i in range(1,6):
        f*=i 
        return f
print(factorial())
```

```python
# Program 19: Random Number
import random
def rand(): 
    return random.randint(1,100)
print(rand())
```

```python
# Program 20: Current Date 

from datetime import date
def today():
    return date.today()
print(today())
```
