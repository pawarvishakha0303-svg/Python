if statement:
An if statement is a decision-making statement in Python that executes a block of code only when the specified condition is True.


```python
 #Check if number is positive
num=int(input("Enter a number"))
if num>0:
    print("Positive number")
```

```python
#Check if number is negative
num=int(input("Enter a number"))
if num<0:
    print("Negative number")
```

```python
#Check if number is Even
num=int(input("Enter a number"))
if num%2==0:
    print("Even number")
```

```python
#Check if number is Odd
num=int(input("Enter a number"))
if num%2!=0:
    print("Odd number")
```

```python
# Check eligible for vote
age=int(input("Enter age:"))
if age>=18:
    print("Eligible for vote")
```
