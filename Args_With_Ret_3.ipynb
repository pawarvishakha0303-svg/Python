```python
# Program 21: Simple Interest

def simple_interest(p,r,t):
    return (p*r*t)/100
print(simple_interest(1000,5,2))
```

```python
# Program 22: Compound Interest 

def compound_interest(p,r,t): 
    return p*((1+r/100)**t)-p
print(round(compound_interest(1000,5,2),2))
```

```python
# Program 23: Area of Circle 

def area_circle(r):
    return 3.14*r*r 
print(area_circle(7))
```

```python
# Program 24: Area of Rectangle 

def area_rectangle(l,b):
    return l*b 
print(area_rectangle(10,5))
```

```python
# Program 25: Celsius to Fahrenheit 

def c_to_f(c): 
    return (c*9/5)+32 
print(c_to_f(37))
```

```python
# Program 26: Fahrenheit to Celsius

def f_to_c(f):
    return (f-32)*5/9 
print(round(f_to_c(98.6),2))
```

```python
# Program 27: Student Grade

def grade(m): 
    return 'A' if m>=90 else 'B' if m>=75 else 'C' if m>=60 else 'D'
print(grade(82))
```

```python
# Program 28: Employee Salary

def net_salary(basic,hra,da):
    return basic+hra+da 
print(net_salary(30000,5000,3000))
```

```python
# Program 29: Shopping Bill

def bill(price,qty):
    return price*qty 
print(bill(250,4))
```

```python
# Program 30: Calculator Addition
def add(a,b):
    return a+b 
print(add(10,20))
```
