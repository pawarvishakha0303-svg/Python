```python
# Merge two dictionaries

d1={"a":1,"b":2,"c":3}
d2={"d":4,"e":5,"f":6}
d1.update(d2)
d1
```

```python
# Print keys and values using loop

d={1:"Yellow",2:"Pink",3:"Black",4:"Orange",5:"White"}
for key,value in d.items():
    print(key,":", value)
```

```python
# Nested dictonary

stud={
      1:{"Name":"Ram","Marks":70},
      2:{"Name":"Neha","Marks":69},
      3:{"Name":"Siya","Marks":85},
      4:{"Name":"Kiran","Marks":90}
}
stud
```

```python
# Sum of dictonary values

d={1:10,2:20,3:30,4:40,5:50}
sum(d.values())
```

```python
# Swap keys and values

d={1:"a",2:"b",3:"c",4:"d",5:"e"}
rev={}
for key,value in d.items():
    rev[value]=key
print(rev)
```

```python
# Sort dictonary by keys

d={"A":20,"C":60,"E":90,"D":50}
for key in sorted(d):
    print(key,d[key])
```

```python
# Print only even values

d={"A":23,"B":20,"C":45,"D":39,"E":50}
for value in d.values():
    if value%2==0:
        print(value)
```

```python
# Print only odd values

d={"A":23,"B":20,"C":45,"D":39,"E":50}
for value in d.values():
    if value%2==1:
        print(value)
```

```python
# Print keys whose values are greater than 50

d={"A":23,"B":55,"C":45,"D":39,"E":50,"F":78,"G":85,"H":98,"I":38}
for key,value in d.items():
    if value>50:
        print(key)
```

```python
## Print keys whose values are less than 50

d={"A":23,"B":55,"C":45,"D":39,"E":50,"F":78,"G":85,"H":98,"I":38}
for key,value in d.items():
    if value<50:
        print(key)
```
