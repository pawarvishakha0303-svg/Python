# With Arguments, No Return Value - 

```python
# Program 1: Factorial

def factorial(n): 
    fact=1 
    for i in range(1,n+1):
        fact*=i 
        print('Factorial =',fact) 
factorial(5) 
```

```python
# Program 2: Prime no 

def prime(n): 
    for i in range(2,n):
        if n%i==0: 
            print('Not Prime');
            return
        print('Prime')
prime(8)
```

```python
# Program 3: Fibonacci 

def fibonacci(n): 
    a,b=0,1 
    for i in range(n):
        print(a,end=' ') 
        a,b=b,a+b 
fibonacci(10) 
```

```python
# Program 4: Reverse String

def reverse(s): 
    print(s[::-1])
reverse('Python')
```

```python
# Program 5: Palindrome 

def palindrome(s): 
    print('Palindrome' if s==s[::-1] else 'Not Palindrome') 
palindrome('madam') 
```

```python
# Program 6: Count Vowels 

def vowels(s):
    c=sum(ch in 'aeiouAEIOU' for ch in s)
    print(c) 
vowels('Python')
```

```python
# Program 7: Sum of Digits 

def sum_digits(n):
    print(sum(map(int,str(n)))) 
sum_digits(1234) 
```

```python
# Program 8: Largest in List
def largest(lst): 
    print(max(lst)) 
largest([2,8,5]) 
```

```python
# Program 9: Average

def average(lst): 
    print(sum(lst)/len(lst)) 
average([10,20,30]) 
```

```python
# Program 10: Remove Duplicates

def remove_dup(lst): 
    print(list(set(lst)))
remove_dup([1,2,2,3])
```
