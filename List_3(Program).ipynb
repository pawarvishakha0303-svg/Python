```python
# Merge Two Lists

list1 = [10, 20, 30]
list2 = [40, 50, 60]

list3 = list1 + list2
print(list3)
```

```python
# Copy a List
list1 = [10, 20, 30]
list2 = list1.copy()

print(list2)
```

```python
# Remove Duplicates
numbers = [10, 20, 10, 30, 20, 40]

unique = list(set(numbers))
print(unique)
```

```python
#: Find Common Elements
list1 = [10, 20, 30, 40]
list2 = [30, 40, 50, 60]

common = []

for i in list1:
    if i in list2:
        common.append(i)

print(common)
```

```python
# Second Largest Element
numbers = [10, 40, 20, 50, 30]

numbers.sort()
print("Second Largest:", numbers[-2])
```

```python
# Separate Even and Odd Numbers
numbers = [10, 15, 20, 25, 30]

even = []
odd = []

for i in numbers:
    if i % 2 == 0:
        even.append(i)
    else:
        odd.append(i)

print("Even:", even)
print("Odd:", odd)
```

```python
# List Comprehension (Square Numbers)
numbers = [1, 2, 3, 4, 5]

square = [i*i for i in numbers]

print(square)
```

```python
# Nested List
matrix = [
    [1, 2],
    [3, 4]
]

print(matrix)
print(matrix[0][1])
```

```python
# Matrix Addition
A = [[1, 2], [3, 4]]
B = [[5, 6], [7, 8]]

result = [[0, 0], [0, 0]]

for i in range(2):
    for j in range(2):
        result[i][j] = A[i][j] + B[i][j]

print(result)
```

```python
# Matrix Multiplication
A = [[1, 2], [3, 4]]
B = [[5, 6], [7, 8]]

result = [[0, 0], [0, 0]]

for i in range(2):
    for j in range(2):
        for k in range(2):
            result[i][j] += A[i][k] * B[k][j]

print(result)
```
