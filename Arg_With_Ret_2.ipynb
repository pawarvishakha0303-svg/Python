```python
# Program 11: Factorial 

def factorial(n): 
    fact=1 
    for i in range(1,n+1):
        fact*=i 
    return fact 
print(factorial(5))
```

```python
# Program 12: Prime Number
def is_prime(n):
    if n<2: 
        return False 
    for i in range(2,n): 
        if n%i==0:
            return False
    return True
print(is_prime(7))
```

```python
# Program 13: Fibonacci List

def fibonacci(n):
    a,b=0,1 
    lst=[]
    for _ in range(n):
        lst.append(a) 
        a,b=b,a+b 
    return lst 
print(fibonacci(10))
```

```python
# Program 14: Reverse Number

def reverse(n):
    return int(str(n)[::-1])
print(reverse(1234))
```

```python
# Program 15: Palindrome Check 

def palindrome(n):
    return str(n)==str(n)[::-1]
print(palindrome(121))
```

```python
# Program 16: Armstrong Check

def armstrong(n): 
    s=sum(int(d)**len(str(n)) 
          for d in str(n))
    return s==n
print(armstrong(153))
```

```python
# Program 17: Perfect Number

def perfect(n): 
    return sum(i for i in range(1,n) if n%i==0)==n 
print(perfect(28))
```

```python
# Program 18: Sum of Digits

def sum_digits(n): 
    return sum(map(int,str(n)))
print(sum_digits(1234))
```

```python
# Program 19: Count Digits

def count_digits(n):
    return len(str(abs(n)))
print(count_digits(12345))
```

```python
# Program 20: GCD 

def gcd(a,b):
    while b: a,b=b,a%b 
    return a
print(gcd(24,36))
```
