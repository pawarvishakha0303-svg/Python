```python
# Program 26: Return Area of Circle

def area_circle():
    r=7 
    return 3.14*r*r 
print(area_circle())
```

```python
# Program 27: Return Area of Rectangle 

def area_rectangle():
    l,b=10,5 
    return l*b
print(area_rectangle())
```

```python
# Program 28: Return Percentage 

def percentage():
    marks=[80,75,90,85,70]
    return sum(marks)/len(marks) 
print(percentage())
```

```python
# Program 29: Return Grade

def grade():
    p=percentage()
    return 'A' if p>=75 else 'B'
print(grade())
```

```python
# Program 30: Return Result Summary

def result_summary(): 
    return {'Percentage':percentage(),'Grade':grade()} 
print(result_summary())
```
