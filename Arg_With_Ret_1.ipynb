```python
# Program 1: Addition 

def add(a,b):
    return a+b 
print(add(10,20))
```

```python
# Program 2: Subtraction

def subtract(a,b):
    return a-b 
print(subtract(20,10))
```

```python
# Program 3: Multiplication

def multiply(a,b):
    return a*b 
print(multiply(5,6))
```

```python
# Program 4: Division

def divide(a,b):
    return a/b if b!=0 else 'Cannot divide by zero' 
print(divide(20,4))
```

```python
# Program 5: Square 

def square(n): 
    return n*n
print(square(7))
```

```python
# Program 6: Cube

def cube(n):
    return n**3
print(cube(4))
```

```python
# Program 7: Power

def power(a,b): 
    return a**b 
print(power(2,5))
```

```python
# Program 8: Maximum 

def maximum(a,b):
    return a if a>b else b 
print(maximum(12,8))
```

```python
# Program 9: Minimum 

def minimum(a,b): 
    return a if a<b else b
print(minimum(12,8))
```

```python
# Program 10: Swap Numbers

def swap(a,b): 
    return b,a
x,y=swap(10,20)
print(x,y)
```
