# Create a set

```python
s={3,6,9,12,15,18,21,24,27,30}
```

```python
state={"Maharashtra","Kerala","Punjab","Karnataka","Bihar"}
```

```python
marks={39,45,50,44,54}
```

A set is a unordered collection,so it does not support indexing

```python
marks[3]
```

# Slicing Support
Sets do not support slicing because they are unordered and have no index positions.

```python
s[2:4]
```
