while loop:
A while loop is used to repeat a block of code as long as a condition is True.

```python
# Name Printing

i=1
while(i<=10):
    print("Neha")
    i=i+1
```

```python
# Print stars(*)

i=1
while i<=5:
    print("*")
    i=i+1
```

```python
# Print Alphabets

i=65
while i<=90:
    print(chr(i))
    i=i+1
```

```python
# Number printing from 1 to 20

i=1
while(i<=20):
    print(i)
    i=i+1
```

```python
# Print 10 to 1

i=10
while i>=1:
    print(i)
    i=i-1
```

```python
# Sum of first 1 to 50 numbers

i=1
total=0
while(i<=50):
    total=total+i
    i=i+1
print("Sum =",total)
```

```python
# Even no. between 1 to 50

i=2
while i<=50:
    print(i)
    i=i+2
```

```python
# Odd no. between 51 to 100

i=51
while i<=100:
    print(i)
    i=i+2
```

```python
# Square from 1 to 10

i=1
while i<=10:
    print(i,"=",i*i)
    i=i+1
```

```python
## Cube from 1 to 10

i=1
while i<=10:
    print(i,"=",i*i*i)
    i=i+1
```
