for Loop:
A for loop is used to repeat a blovk of code for each item in a sequence.

```python
# Numbers from 1 to 10

for i in range(1,11,1):
    print(i)
```

```python
# Numbers from 10 to 1

for i in range(10,0,-1):
    print(i)
```

```python
# Even numbers from 1 to 20

for i in range(2,21,2):
    print(i)
    
```

```python
# Odd numbers from 1 to 20

for i in range(1,21,2):
    print(i)
```

```python
# Square of numbers from 1 to 10

for i in range(1,11):
    print(i,"=",i*i)
```

```python
# Print name n times

for i in range(10):
    print("Shreya")
```

```python
# Each character of string

name="python"

for ch in name:
    print(ch)
```

```python
# Print small Alphabet

for i in range(97,123):
    print(chr(i))
```

```python
# Pattern Printing

for i in range(1,5,1):
    print('* '*i)
```

```python
for i in range(5):
    print("# "*5)
```
