IF_ELSE Statememt:
An if-else statement is a decision-making statement in Python. It checks a condition:
If the condition is True, the code inside the if block executes.
If the condition is False, the code inside the else block executes.


```python
# Check number is greater or not

n1=23
n2=90
if(n1>n2):
    print(f"{n1} grater than {n2}")

else:
    print(f"{n2} greater than {n1}")
    
```

```python
# Check given number is odd or even

n1=int(input("Enter a number"))
if n1%2==1:
    print("Number is Odd")
else:
    print("Number is Even")
    
```

```python
# Check number is positive or negative

n=int(input("Enter a number:"))
if n>0:
    print("Number is Positive")
else:
    print("Number is Negative")
```

```python
# Check age is eligible for vote

age=int(input("Enter Age:"))
if age>=18:
    print("Eligible for Vote")
else:
    print("Not eligible for Vote")
```

```python
# Check year is leap or not

year=int(input("Enter Year"))
if year % 4 == 0:
    print("Leap year")
else:
    print("Not a Leap year")
```

```python
# Check number is divisible by 5

num=int(input("Enter number:"))
if num%5==0:
    print("Number is divisible by 5")
else:
    print("Number is not divisible by 5")
```

```python
# Check alphabet or not

ch=input("Enter a character:")
if ch.isalpha():
    print("Alphabet")
else:
    print("Not an Alphabet")
```

```python
#Check pass or fail

marks=int(input("Enter marks:"))
if marks>=35:
    print("Pass")
else:
    print("Fail")
                
                
```
