# No Arguments, No Return Value – Basic Programs 
✅ No parameters are passed to the function.

✅ The function does not return any value using return.

✅ The function performs its task by printing the output directly.

```python
# Program 1: Print Hello World

def hello():
    print("Hello, World!")

hello()
```

```python
# Program 2: Print Welcome Message

def welcome():
    print("Welcome to Python!")

welcome()
```

```python
# Program 3: Print Your Name

def my_name():
    print("My name is Vishakha.")

my_name()
```

```python
# Program 4: Print Numbers from 1 to 10

def print_numbers():
    for i in range(1, 11):
        print(i)

print_numbers()
```

```python
# Program 5: Print Even Numbers from 1 to 20

def even_numbers():
    for i in range(2, 21, 2):
        print(i)

even_numbers()
```

```python
# Program 6: Print Odd Numbers from 1 to 20

def odd_numbers():
    for i in range(1, 21, 2):
        print(i)

odd_numbers()
```

```python
# Program 7: Print Multiplication Table of 5

def table():
    for i in range(1, 11):
        print(f"5 x {i} = {5 * i}")

table()
```

```python
# Program 8: Print Square of Numbers (1 to 10)

def squares():
    for i in range(1, 11):
        print(i, "=", i ** 2)

squares()
```

```python
# Program 9: Print Cube of Numbers (1 to 10)

def cubes():
    for i in range(1, 11):
        print(i, "=", i ** 3)

cubes()
```

```python
# Program 10: Print Alphabet A to Z

def alphabets():
    for ch in range(65, 91):
        print(chr(ch), end=" ")

alphabets()
```
