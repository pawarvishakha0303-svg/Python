# lambda function:
A lambda function in Python is a small, anonymous (unnamed) function that can have any number of arguments but only one expression.

```python
# Program 1 :Square of number

square = lambda x: x * x

print(square(5))
```

```python
# Program 2 :add_two_numbers.py

add = lambda a, b: a + b


print(add(10, 20))

```

```python
# Program 3 : cube

cube = lambda x: x ** 3

print(cube(4))
```

```python
# Program 4 : Addition of two numbers

add = lambda a, b: a + b

print(add(10, 20))
```

```python
# Program 5 : Multiplication of two numbers

multiply = lambda a, b: a * b

print(multiply(5, 4))
```

```python
# Program 6 : Division

divide = lambda a, b: a / b

print(divide(20, 5))
```

```python
# Program 7 : Check even no

even = lambda x: x % 2 == 0

print(even(10))
```

```python
# Program 8 : Check Odd number

odd = lambda x: x % 2 != 0

print(odd(7))
```

```python
# Program 9 : Find maximum

maximum = lambda a, b: a if a > b else b

print(maximum(15, 25))
```

```python
# Program 10 : Check positive or negative

check = lambda x: "Positive" if x > 0 else "Negative"

print(check(-5))
```
