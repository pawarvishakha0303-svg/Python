# File Handling
File handling means creating, reading, writing, updating, and deleting files using Python.

```python
# Program 1: Create a File

file = open("sample.txt", "w")
file.close()

print("File created successfully")
```

```python
# Program 2:Write Data to a File

file = open("sample.txt", "w")

file.write("Hello Python")

file.close()

print("Data written successfully")
```

```python
# Program 3:Read Data from a File

file = open("sample.txt", "r")

data = file.read()

print(data)

file.close()
```

```python
# Program 4: Read the First Line

file = open("sample.txt", "r")

data = file.readline()

print(data)

file.close()
```

```python
# Program 5:Read All Lines

file = open("sample.txt", "r")

data = file.readlines()

print(data)

file.close()
```

```python
# Program 6:Append Data to a File

file = open("sample.txt", "a")

file.write("\nWelcome to Python")

file.close()

print("Data appended successfully")
```

```python
# Program 7:Count Characters in a File

file = open("sample.txt", "r")

data = file.read()

print("Characters:", len(data))

file.close()
```

```python
# Program 8: Count Words in a File

file = open("sample.txt", "r")

data = file.read()

words = data.split()

print("Words:", len(words))

file.close()
```

```python
# Program 9: Count Lines in a File

file = open("sample.txt", "r")

data = file.readlines()

print("Lines:", len(data))

file.close()
```

```python
# Program 10: Check if a File Exists

import os

if os.path.exists("sample.txt"):
    print("File exists")
else:
    print("File does not exist")
```
