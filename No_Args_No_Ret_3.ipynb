```python
# Program 1. Star Triangle

def program():
    for i in range(1, 6):
        print("*" * i)

program()
```

```python
# Program 2: Number Triangle

def program():
    for i in range(1, 6):
        for j in range(1, i + 1):
            print(j, end=" ")
        print()

program()
```

```python
# Program 3: Decreasing number

def program():
    for i in range(5, 0, -1):
        for j in range(i):
            print(i, end=" ")
        print()

program()
```

```python
# Program 4. Solid Square

def program():
    n = 5

    for i in range(n):
        print("* " * n)

program()
```

```python
# Program 5: Rectangle

def program():
    rows = 4
    columns = 8

    for i in range(rows):
        print("* " * columns)

program()
```

```python
# Program 6: Alphabet triangle

def program():
    for i in range(1, 6):
        for j in range(i):
            print(chr(65 + j), end=" ")
        print()
program()
```

```python
# Program 7:Repeated alphabet

def program():
    for i in range(5):
        for j in range(i + 1):
            print(chr(65 + i), end=" ")
        print()

program()
```

```python
# Program 8: Box number

def program():
    n = 5

    for i in range(1, n + 1):
        for j in range(1, n + 1):
            print(j, end=" ")
        print()

program()
```

```python
# Program 9: Repeated number triangle

def program():
    for i in range(1, 6):
        for j in range(i):
            print(i, end=" ")
        print()

program()
```

```python
# Program 10: Reverse alphabet

def program():
    for i in range(5, 0, -1):
        for j in range(i):
            print(chr(65 + j), end=" ")
        print()

program()
```
