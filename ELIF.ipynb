ELIF Statement:
elif is used to check another condition when the previous if condition is False.


```python
# Greater number

n1=78
n2=34

if n1>n2:
    print(f"{n1} grater than {n2}")
    
elif(n2>n1):
    print(f"{n2} greater than {n1}")
    
else:
    print(f"{n1} and {n2} are equal")
    
```

```python
# Check Positive ,Negative or Zero

num=int(input("Enter a number"))
if num>0:
    print("Positive Number")
elif num<0:
    print("Negative Number")
else:
    print("Zero")
```

```python
# Check character type

ch= input("Enter a character:")
if ch.isalpha():
    print("Alphabet")
elif ch.isdigit():
    print("Digit")
else:
    print("Special Character")
```

```python
# Marks

chem=int(input("Enter marks of chem:"))
bio=int(input("Enter marks of bio: "))
maths=int(input("Enter marks maths: "))
phy=int(input("Enter marks of phy:"))
stat=int(input("Enter marks stat: "))

total=chem+bio+maths+phy+stat

per=total/500*100

print(f"Total marks is {total}:")
print(f"Percentage is: {per}")
if(per>75):
    print("Greade A")
elif(per>50):
    print("Grade B")
elif(per>35):
    print("Grade C")
else:
    print("Fail")
```
