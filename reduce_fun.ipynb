# redude() :
The reduce() function is used to reduce a sequence of values to a single value by repeatedly applying a function.

It is available in the functools module.

```python
# Program 1 : Sum of Numbers

from functools import reduce

numbers = [1, 2, 3, 4, 5]

result = reduce(lambda x, y: x + y, numbers)

print(result)
```

```python
# Program 2 : Product of Numbers

from functools import reduce

numbers = [1, 2, 3, 4, 5]

result = reduce(lambda x, y: x * y, numbers)

print(result)
```

```python
# Program 3 : Find Maximum

from functools import reduce

numbers = [10, 25, 15, 40, 30]

result = reduce(lambda x, y: x if x > y else y, numbers)

print(result)
```

```python
# Program 4 : Find Minimum

from functools import reduce

numbers = [10, 25, 15, 40, 30]

result = reduce(lambda x, y: x if x < y else y, numbers)

print(result)
```

```python
# Program 5 : Sum of Even Numbers

from functools import reduce

numbers = [1, 2, 3, 4, 5, 6]

even_numbers = filter(lambda x: x % 2 == 0, numbers)

result = reduce(lambda x, y: x + y, even_numbers)

print(result)
```

```python
# Program 6 : Sum of Odd Numbers

from functools import reduce

numbers = [1, 2, 3, 4, 5, 6]

odd_numbers = filter(lambda x: x % 2 != 0, numbers)

result = reduce(lambda x, y: x + y, odd_numbers)

print(result)
```

```python
# Program 7 : Multiply All Numbers

from functools import reduce

numbers = [2, 3, 4]

result = reduce(lambda x, y: x * y, numbers)

print(result)
```

```python
# Program 8 : Find Longest Word

from functools import reduce

words = ["Python", "Java", "Programming", "Code"]

result = reduce(lambda x, y: x if len(x) > len(y) else y, words)

print(result)
```

```python
# Program 9 : Combine Strings

from functools import reduce

words = ["Python", "is", "easy"]

result = reduce(lambda x, y: x + " " + y, words)

print(result)
```

```python
# Program 10 : Calculate Factorial

from functools import reduce

number = 5

result = reduce(lambda x, y: x * y, range(1, number + 1))

print(result)
```
