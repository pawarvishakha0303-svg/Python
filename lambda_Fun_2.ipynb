```python
# Program 1 : Find Length of a String

length = lambda text: len(text)

print(length("Python"))
```

```python
# Program 2:  Convert String to Uppercase

upper = lambda text: text.upper()

print(upper("python"))
```

```python
# Program 3: Convert String to Lowercase

lower = lambda text: text.lower()

print(lower("PYTHON"))
```

```python
# Program 4: Find First Character

first = lambda text: text[0]

print(first("Python"))
```

```python
# Program 5: Find Last Character

last = lambda text: text[-1]

print(last("Python"))
```

```python
# Program 6: Reverse a String

reverse = lambda text: text[::-1]

print(reverse("Python"))
```

```python
# Program 7: Check Positive, Negative, or Zero

check = lambda x: "Positive" if x > 0 else "Negative" if x < 0 else "Zero"

print(check(0))
```

```python
# Program 8: Calculate Simple Interest

interest = lambda p, r, t: (p * r * t) / 100

print(interest(10000, 5, 2))
```

```python
# Program 9: Calculate Area of a Circle

area = lambda r: 3.14 * r * r

print(area(5))
```

```python
# Program 10: Find Smallest of Three Numbers

smallest = lambda a, b, c: min(a, b, c)

print(smallest(10, 25, 5))
```
