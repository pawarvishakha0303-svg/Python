```python
# Simple shopping bill

print("===== SHOPPING BILL =====")

item1 = input("Enter item 1: ")
price1 = float(input("Enter price: "))
quantity1 = int(input("Enter quantity: "))

item2 = input("Enter item 2: ")
price2 = float(input("Enter price: "))
quantity2 = int(input("Enter quantity: "))

item3 = input("Enter item 3: ")
price3 = float(input("Enter price: "))
quantity3 = int(input("Enter quantity: "))

total1 = price1 * quantity1
total2 = price2 * quantity2
total3 = price3 * quantity3

total = total1 + total2 + total3

if total >= 1000:
    discount = total * 0.10
else:
    discount = 0

final_amount = total - discount

print("\n===== BILL =====")
print(item1, "x", quantity1, "=", total1)
print(item2, "x", quantity2, "=", total2)
print(item3, "x", quantity3, "=", total3)

print("-------------------------")
print("Total:", total)
print("Discount:", discount)
print("Final Amount:", final_amount)
print("-------------------------")
print("Thank you for shopping!")
```
