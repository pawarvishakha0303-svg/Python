# Part 3: Advanced String Interview Programs 


```python
# Program 21: Remove Duplicate Characters

s = input("Enter a string: ")
result = ""

for ch in s:
    if ch not in result:
        result += ch

print("After removing duplicates:", result)
```

```python
# Program 22: Find the Most Frequent Character

s = input("Enter a string: ")

max_char = ""
max_count = 0

for ch in s:
    count = s.count(ch)
    if count > max_count:
        max_count = count
        max_char = ch

print("Most frequent character:", max_char)
print("Frequency:", max_count)
```

```python
# Program 23: Reverse Each Word in a Sentence

s = input("Enter a sentence: ")

words = s.split()

for word in words:
    print(word[::-1], end=" ")
```

```python
# Program 24: Reverse the Order of Words

s = input("Enter a sentence: ")

words = s.split()

print(" ".join(words[::-1]))
```

```python
# Program 25: Find All Duplicate Characters

s = input("Enter a string: ")

printed = ""

for ch in s:
    if s.count(ch) > 1 and ch not in printed:
        print(ch)
        printed += ch
```

```python
# Program 26: Find the First Repeated Character

s = input("Enter a string: ")

seen = ""

for ch in s:
    if ch in seen:
        print("First repeated:", ch)
        break
    seen += ch
```

```python
# Program 27: Find the First Non-Repeated Character

s = input("Enter a string: ")

for ch in s:
    if s.count(ch) == 1:
        print("First non-repeated:", ch)
        break
```

```python
# Program 28: Count Frequency of Each Word

s = input("Enter a sentence: ")

words = s.split()

for word in set(words):
    print(word, ":", words.count(word))
```

```python
# Program 29: Find Shortest Word

s = input("Enter a sentence: ")

words = s.split()

shortest = min(words, key=len)

print("Shortest word:", shortest)
```

```python
# Program 30: Find Longest Word

s = input("Enter a sentence: ")

words = s.split()

longest = max(words, key=len)

print("Longest word:", longest)
```
