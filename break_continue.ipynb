break_continue:

```python
# Skip 5 and stop at 8

for i in range(1, 11):
    if i == 5:
        continue
    if i == 8:
        break
    print(i)
```

```python
# Skip 8 and stop at 10

for i in range(1, 11):
    if i == 8:
        continue
    if i == 10:
        break
    print(i)
```

```python
# Skip even numbers and stop at 9

for i in range(1, 11):
    if i % 2 == 0:
        continue
    if i == 9:
        break
    print(i)
```

```python
# Print Odd Numbers, Stop at 19

for i in range(1, 21):
    if i % 2 == 0:
        continue
    if i == 19:
        break
    print(i)
```

```python
# Skip negative numbers and stop when 100 is entered

for i in range(5):
    n = int(input("Enter a number: "))

    if n < 0:
        continue

    if n == 100:
        break

    print(n)
```

```python
# Skip spaces and stop at .

text = input("Enter text: ")

for ch in text:
    if ch == " ":
        continue
    if ch == ".":
        break
    print(ch)
```

```python
# Skip multiples of 3 and stop at 20

for i in range(1, 21):
    if i % 3 == 0:
        continue
    if i == 20:
        break
    print(i)
```

```python
# Skip number 4 stop at 8

for i in range(1, 11):

    
    if i == 4:
        continue      # Skip number 4

    if i == 8:
        break         # Stop the loop at 8

    print(i)
```
