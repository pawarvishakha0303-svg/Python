continue statement:
Skips the current iteration and continues with the next iteration.

```python
# Skip no. 5

for i in range(1, 11):
    if i == 5:
        continue
    print(i)
```

```python
# Skip even no

for i in range(1, 21):
    if i % 2 == 0:
        continue
    print(i)
```

```python
# Skip Odd no.

for i in range(21, 51):
    if i % 2 == 1:
        continue
    print(i)
```

```python
# Skip letter a

word = input("Enter a word: ")

for ch in word:
    if ch == "a":
        continue
    print(ch)
```

```python
# Skip letter h

word = input("Enter a word: ")

for ch in word:
    if ch == "h":
        continue
    print(ch)
```

```python
# Skip multiples of 3

for i in range(1, 16):
    if i % 3 == 0:
        continue
    print(i)
```

```python
# Skip number greater than 5

for i in range(1, 11):
    if i > 5:
        continue
    print(i)
```

```python
#skip number divisible by 4

for i in range(1, 21):
    if i % 4 == 0:
        continue
    print(i)
```

```python
# Skip positive numbers

numbers = [-5, 8, -2, 10, -7]

for num in numbers:
    if num > 0:
        continue
    print(num)
```

```python
# Skip number end with 5

for i in range(1, 31):
    if i % 10 == 5:
        continue
    print(i)
```
