break Statement:
Is is used to immediately terminate(exit) a loop when a certain condition is met

```python
# stop loop when i=5

for i in range(1,11):
    if i==5:
        break
    print(i)
    
```

```python
# stop loop when i=10

for i in range(1,21):
    if i==10:
        break
    print(i)
```

```python
# Stop when a character is found

name=input("Enter a name")
for ch in name:
    if ch=="a":
        break
    print(ch)
```

```python
# Stop at first multiple of 5

for i in range(1,21):
    if i%5==0:
        break
    print(i)
```

```python
# Stop when number is zero

num = int(input("Enter a number:"))
for i in range(5):
    if num==0:
        break
    print(num)
    
```

```python
# Stop after printing 10 numbers

for i in range(1,21):
    if i==10:
        break
    print(i)
```

```python
# Stop when number is greater than 100

for i in range(5):
    num=int(input("Enter a number:"))
    if num>100:
        break
    print(num)
```

```python
# Stop when marks below 35


for i in range(5):
    marks=int(input("Enter Marks:"))
    if marks<35:
        break
    print(marks)
```

```python
# Stop when ahe is 18 or above

for i in range(5):
    age=int(input("Enter Age:"))
    if age>=18:
        print("Eligible for vote")
    break
```
