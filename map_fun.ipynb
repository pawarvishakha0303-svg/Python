# map() :
The map() function is used to apply the same function to every item in an iterable such as a list or tuple.

```python
# Program 1 : Double each number

numbers = [1, 2, 3, 4, 5]

result = map(lambda x: x * 2, numbers)

print(list(result))
```

```python
# Program 2 : Square each number

numbers = [1, 2, 3, 4, 5]

result = map(lambda x: x ** 2, numbers)

print(list(result))
```

```python
# Program 3 : Cube each number

numbers = [1, 2, 3, 4, 5]

result = map(lambda x: x ** 3, numbers)

print(list(result))
```

```python
# Program 4 : Add 10 to each number

numbers = [1, 2, 3, 4, 5]

result = map(lambda x: x + 10, numbers)

print(list(result))
```

```python
# Program 5 :Convert strings to uppercase

names = ["vishakha", "python", "github"]

result = map(str.upper, names)

print(list(result))
```

```python
# Program 6 : Convert strings to lowercase

names = ["PYTHON", "GITHUB", "PROGRAM"]

result = map(str.lower, names)

print(list(result))
```

```python
# Program 7 : Find length of each word

words = ["Python", "GitHub", "Code"]

result = map(len, words)

print(list(result))
```

```python
# Program 8 : Convert strings to integers

numbers = ["10", "20", "30", "40"]

result = map(int, numbers)

print(list(result))
```

```python
# Program 9 : Add two lists

a = [1, 2, 3, 4]
b = [10, 20, 30, 40]

result = map(lambda x, y: x + y, a, b)

print(list(result))
```

```python
# Program 10 : Multiply two lists

a = [1, 2, 3, 4]
b = [2, 3, 4, 5]

result = map(lambda x, y: x * y, a, b)

print(list(result))
```
